```python
import json
from datetime import datetime

def convert_cookie_string_to_json(cookie_str, filename="instagram_cookies.json"):
    """
    将cookie字符串转换为Selenium可用的JSON格式
    
    Args:
        cookie_str: 原始的cookie字符串
        filename: 输出文件名
    """
    # 从cookie字符串中提取cookies
    cookies = {}
    
    # 清理字符串
    cookie_str = cookie_str.strip().replace("\\n", "")
    
    # 查找cookie部分
    if "cookie: " in cookie_str.lower():
        # 提取cookie部分
        cookie_line = ""
        for line in cookie_str.split('\\n'):
            line = line.strip()
            if line.lower().startswith("cookie: "):
                cookie_line = line[8:].strip()  # 去掉"cookie: "
                break
        
        if cookie_line:
            # 解析各个cookie
            cookie_items = cookie_line.split("; ")
            for item in cookie_items:
                if "=" in item:
                    name, value = item.split("=", 1)
                    cookies[name] = value
    
    # 如果没有找到cookie: 格式，尝试其他方式
    if not cookies:
        # 直接处理整个字符串
        cookie_items = cookie_str.split("; ")
        for item in cookie_items:
            if "=" in item:
                name, value = item.split("=", 1)
                cookies[name] = value
    
    # 转换为Selenium cookie格式
    selenium_cookies = []
    domain = "www.instagram.com"
    
    for name, value in cookies.items():
        cookie_dict = {
            "name": name,
            "value": value,
            "domain": ".instagram.com",
            "path": "/",
            "secure": True,
            "httpOnly": False,
            "sameSite": "Lax"
        }
        
        # 针对特定cookie设置
        if name == "c_user":
            cookie_dict["httpOnly"] = False
        elif name in ["xs", "fr"]:
            cookie_dict["httpOnly"] = True
            cookie_dict["secure"] = True
        
        selenium_cookies.append(cookie_dict)
    
    # 保存到文件
    with open(filename, 'w', encoding='utf-8') as f:
        json.dump(selenium_cookies, f, indent=2, ensure_ascii=False)
    
    print(f" 已转换 {len(selenium_cookies)} 个cookie到 {filename}")
    print("转换的cookie列表:")
    for cookie in selenium_cookies:
        print(f"  - {cookie['name']}: {cookie['value'][:20]}...")
    
    return selenium_cookies

# 使用示例
if __name__ == "__main__":
    # 你的cookie字符串
    cookie_string = """
datr=WPFlabOy6DNAn4_mpuHUntbD; ig_did=99FB3D16-50EE-4652-A6A9-0B68ABF76914; dpr=1.5; mid=aWXxWAALAAHquuy89IoMOODZjmNf; ig_nrcb=1; fbm_124024574287414=base_domain=.instagram.com; csrftoken=urs4kCKHNiX65mQ0JoGncVBwIiKhl43h; ds_user_id=80029964293; wd=638x559; sessionid=80029964293%3AWJ9YzOgZwypZ32%3A13%3AAYjZZNQqHmrE1obS6XUb2-h9XXscilcELgC8tijtDQ; ps_l=1; ps_n=1;"""
    
    # 转换并保存
    convert_cookie_string_to_json(cookie_string)
```
